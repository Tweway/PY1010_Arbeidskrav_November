## Oppgave 2 – Pizzaberegning

Det skal arrangeres en klassefest, og man antar at hver elev spiser 1/4 pizza.

Lag et program som tar inn antall elever fra konsollen ved:

```python
antall_elever = int(input('Skriv inn antall elever: '))
```

Programmet skal så regne ut hvor mange pizzaer som skal handles inn til festen og skrive svaret til skjerm.

**Merk:** Man kan ikke kjøpe 4 og en kvart pizza på butikken (man må da kjøpe 5).

### Hint 1
Gi programmet ditt et fornuftig svar hvis det for eksempel er 21 elever i klassen.

### Hint 2
Det er ikke vanlig å skrive:

«Det må handles inn 6.0 pizzaer til festen».

Hvordan kan du sikre at antall pizzaer skrives ut som et heltall (ikke desimaltall)?

````python
"""
  --- Pizza Program ---

  Under funker koden ganske likt som oppgave 1. Her er det en mattefunksjon under som
  regner ut antall 5 kvarts pizzaer som deles på 21 stykker og, og tilnærmer ned
  slik at det ikke blir altfor skarpt svar men også sier hvor mye reste stykker.

  Tillegg til å gjøre koden bruker vennlig med heltall og enkelt antall.
  Selve "antall_elever" er bygd likt som spørsmål koden i opg. 1
  Regnestykke i seg selv er elevene ganger stykke pizzer de spiser hver 
  som deles på totalt pizza stykker i en pizza. 

Tok meg også friheten her til å bruke f string siden jeg syns det er ryddigere og 
for å øve inn til mer komplekse koder. Var usikker på om det var noe regel til komplisitet
så bruker også "import math" og math ceiling og floor til å runde opp og ned. Siden oppgaven 
spurte om heltall men da såklart ikke bare fjerne desimalene med " .0f "
"""


import math # for å kjøre avrunding

antall_elever = int(input("Skriv inn antall elever: "))

antall_pizzaslices_pr_pers = 0.25 #som er 1/4 av en pizza 
antall_pizzaslices_pr_pizza= 5
antall_pizzaer = math.ceil(antall_elever * antall_pizzaslices_pr_pers)  #som her regner antall elever ganger hvor mye av en pizza de spiser
antall_pizza_leftover =math.floor((antall_pizzaslices_pr_pizza * antall_pizzaer) - (antall_elever * antall_pizzaslices_pr_pers * antall_pizzaslices_pr_pizza))
#Tok friheten til å runde ned på restene enn å tenke an får mer enn man faktisk får. Så det heller blir bonus. Alternativt kunne ha latt være å la det stå 2.3 slizes.
#pga dette kan det bli forkastet små deler som ikke er hele pizza biter. Math floor viser dermed i min kode hvor mange hele stykker

print(f"Dere vil trenge å handle inn {antall_pizzaer} pizzaer for at alle {antall_elever} av dere får 1/4 pizza hver. Det vil også bli {antall_pizza_leftover} hele stykker igjen etter alle får likt")
````

## Oppgave 5 – Funksjon og geometri

Lag et program med en funksjon som tar `a` og `b` som inn-argumenter og som regner ut arealet og «ytre» omkrets til en figur satt sammen av en rettvinklet trekant og en halvsirkel, se figuren under.

Med «ytre» omkrets menes samlet lengde av de sorte strekene.

Funksjonen skal returnere arealet og «ytre» omkrets, som så skrives til skjerm med passende tekst.

### Formler fra figuren

**Halvsirkel:**

- Areal av sirkel: se bilde.
- Omkrets av sirkel: se bilde.

**Trekant:**

- Areal av trekant: se bilde.
- Omkrets av trekant: se bilde.

**Merk:** Figuren består av en rettvinklet trekant og en halvsirkel. Den indre/skjulte kanten mellom figurene skal ikke regnes med i den ytre omkretsen.


````python
#Oppgave 5 
import numpy as np  
def figur(a, b):
    radius_a = a / 2 #dette er for sikrelen
    areal_a = (a*b) / 2
    areal_halvsirkel = (np.pi * radius_a**2) / 2
    areal = areal_a + areal_halvsirkel

    c = np.sqrt(a**2 + b**2) #denne er for trekanten
    bue=np.pi * radius_a
    omkrets = b + c + bue
    return areal, omkrets

areal, omkrets = figur(4, 3) #4 og 3 som test eksempler her hvor første er a verdi og andre er b verdien
print(f"Arealet av figuren er {areal:.2f} og omkretsen er {omkrets:.2f}")
````
